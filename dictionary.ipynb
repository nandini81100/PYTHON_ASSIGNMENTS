```python
# Question 50: Find union of all nested sets:
my_dict = {
    'groups': [{1, 2, 3}, {4, 5, 6}],
    'categories': [{'a', 'b'}, {'c', 'd'}]
}

# Find union of groups
groups_union = my_dict['groups'][0].union(my_dict['groups'][1])

# Find union of categories
categories_union = my_dict['categories'][0].union(my_dict['categories'][1])

print("Groups union:", groups_union)
print("Categories union:", categories_union)
```
