```python
# 97 is prime number or not: 6
num = 97

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print( " 97 is a Prime number")
else:
    print("97 is Not a prime number")
```
