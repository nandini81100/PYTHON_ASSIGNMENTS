```python
# Question 23: Convert "hello world" to "hElLo WoRlD" (alternating case)
text = "hello world"

result = ""
for i, char in enumerate(text):
    if char != " ":
        if len(result.replace(" ", "")) % 2 == 0:
            result += char.lower()
        else:
            result += char.upper()
    else:
        result += char

print(result)

```

```python
text = "hello world"

morse = {
    "h": "....", "e": ".", "l": ".-..", "o": "---",
    "w": ".--", "r": ".-.", "d": "-.."
}

result = ""

for char in text:
    if char == " ":
        result += " "
    else:
        result += morse[char] + " "

print(result)
```

```python
str1 = "programming"
str2 = "grammar"

longest = ""

for i in range(len(str1)):
    for j in range(i + 1, len(str1) + 1):
        substring = str1[i:j]

        if substring in str2 and len(substring) > len(longest):
            longest = substring

print(longest)
```
