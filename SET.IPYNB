```python
# Question 1: Create a set of first 10 natural numbers:

set_1=set(range(1,11))
print(set_1)
```

```python
# Question 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}:

numbers={1, 2, 3, 4, 5, 6}

numbers.remove(2)
numbers.remove(6)
numbers.remove(3)
print(numbers)
```

```python

# Question 4: Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8} :
set_1={1, 2, 3, 4, 5}
set_2={4, 5, 6, 7, 8}
new_set=set_1.intersection(set_2)
print(new_set)
```

```python
# Question 5: Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}:

set_1={1, 2, 3, 4, 5} 
set_2={4, 5, 6, 7, 8}
new_set=set_1.difference(set_2)
print(new_set)
```

```python
# Question 6: Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}

set_1 = {1, 2, 3, 4, 6, 7, 8}

if 5 in set_1:
    print("5 is present in the set")
else:
    print("5 is not present in the set")
```

```python
# Question 7: Find the length of set {'a', 'b', 'c', 'd', 'e'}:

set_1={'a', 'b', 'c', 'd', 'e'}
length_1=len(set_1)
print(length_1)
```

```python
# Question 8: Create a set of vowels from string "hello world"

text = "hello world"
vowels = set()

for char in text:
    if char in "aeiou":
        vowels.add(char)

print(vowels)
```

```python
# Question 9: Remove duplicates from list using set

numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6, 7]
nonduplicate_numbers=set(numbers)
print(nonduplicate_numbers)
```

```python
# Question 10: Check if {1, 2, 3} is a subset of {1, 2, 3, 4, 5, 6}:



set_1 = {1, 2, 3}
set_2 = {1, 2, 3, 4, 5, 6}

if set_1.issubset(set_2):
    print("set_1 is a subset of set_2")
else:
    print("set_1 is not a subset of set_2")

```
