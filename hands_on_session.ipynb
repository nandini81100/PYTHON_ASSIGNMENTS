```python
# Right-angled triangle — stars:

n=5
for i in range (1,n+1):
  print("*" * i)
```

```python
 #Inverted right-angled triangle — stars:

n=5
for i in range(n,0,-1):
  print("*" * i)
```

```python
# Solid square of stars:
n=5
for _ in range(n):
   print("*" * n)
```

```python
#hollow sqaure:

n=5
for i in range(n):
  if i==0 or i== n - 1:
    print("*" * n)
  else:
    print("*" + " " * (n-2) + "*")
```

```python
#.5 Number triangle (1, 12, 123, …):
n=5
for i in range(1, n +1):
  for j in range(1, i +1):
    print(j,end= " ")

  print()
```

```python
#6. Floyd’s triangle (continuous numbers):
n=4
num=1
for i in range(1,n+1):
  for j in range (1, i+1):
    print(num,end=" ")
    num+=1
  print()
```

```python
#7. Centered star pyramid:
n=4
for i in range(1,n + 1):
  spaces=" " * (n-i)
  stars="*" * (2 * i -1)
  print(spaces+ stars)
```

```python
#8 inverted star pyramid:
n=5
for i in range(5,0,-1):
  spaces=" " *(n-i)
  stars ="*" * (2 *i-1)
  print(spaces+stars)
```

```python
#9. Diamond (centered, stars):

n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```
