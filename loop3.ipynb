```python
#Problem 1.1: Grade Calculator (Easy)
#Write a program that takes a student's marks (0-100) as input and prints their grade according to the following criteria:


marks = int(input(" enter your marks here  "))
if marks >=90:
   print("GRADE"  "A")
elif marks >=80:
  print("GRADE  ""B")
elif marks>=70:
  print("GRADE"  "C")
elif marks >=60:
  print("GRADE" "D")
else:
  print("GRADE"  "F")

```

```python
#Problem 1.4: BMI Calculator with Categories (Easy):
weight = int(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

if weight <= 0 or height <= 0:
    print("Invalid weight or height")

else:
    BMI = weight / (height ** 2)

    print("Your BMI is:", round(BMI, 2))

    if BMI < 18.5:
        print(" you are Underweight")

    elif BMI < 25:
        print("Normal weight")

    elif BMI< 30:
        print("Overweight")

    else:
        print("Obese")
```
